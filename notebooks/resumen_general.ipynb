```python
df_users = pd.read_csv('../data/tablas-actualizadas/users.csv')
df_users.shape, df_users.columns
```

```python
df_devices = pd.read_csv('../data/tablas-actualizadas/devices.csv')
df_devices.shape, df_devices.columns
```

```python
df_notifications = pd.read_csv('../data/tablas-actualizadas/notifications.csv')
df_notifications.shape, df_notifications.columns
```

```python
df_transactions = pd.read_csv('../data/tablas-actualizadas/transactions.csv')
df_transactions.shape, df_transactions.columns
```

```python
for df in [df_users, df_devices, df_notifications, df_transactions]:
    if 'Unnamed: 0' in df.columns:
        df.drop(columns='Unnamed: 0', inplace=True)
```

```python
df_users.rename(columns={
    'created_date': 'create_date_user'
}, inplace=True)
```

```python
df_devices.rename(columns={
    'string_field_0': 'device_type',
    'string_field_1': 'user_id'
}, inplace=True)
```

```python
df_notifications.rename(columns={
    'created_date': 'create_date_notification'
}, inplace=True)
```

```python
df_transactions.rename(columns={
    'created_date': 'create_date_transaction'
}, inplace=True)
```

```python
df_users.columns, df_devices.columns, df_notifications.columns, df_transactions.columns
```

```python
import pandas as pd

# Convertir fechas en ambos DataFrames
df_users['create_date_user'] = pd.to_datetime(df_users['create_date_user'], errors='coerce')
df_transactions['create_date_transaction'] = pd.to_datetime(df_transactions['create_date_transaction'], errors='coerce')

# Merge básico: unir solo por user_id, manteniendo TODAS las columnas originales de ambos
df_merged_RG = pd.merge(
    df_users,
    df_transactions,
    on='user_id',
    how='left',             # para que estén todos los usuarios aunque no tengan transacciones
    suffixes=('_user', '_txn')
)

# Verificar columnas resultantes
print(f"Columnas en df_merged_RG: {df_merged_RG.columns.tolist()}")

# Ver las primeras filas para inspección
df_merged_RG.head(), df_merged_RG.shape
```

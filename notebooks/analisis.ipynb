# Importando tablas

```python
df_devices = pd.read_csv('../data/devices.csv')
df_notifications = pd.read_csv('../data/notifications.csv')
df_transactions = pd.read_csv('../data/transactions.csv')
df_users = pd.read_csv('../data/users.csv')

display(df_devices.head(5), df_notifications.head(5), df_transactions.head(5), df_users.head(5))
```

```python
display(df_devices.info(), df_notifications.info(), df_transactions.info(), df_users.info())
```

# Rename

```python
## cambiar el nombre a la tabla

df_devices = df_devices.rename(columns={"string_field_0": "brand_device", "string_field_1": "user_id"})
df_notifications = df_notifications.rename(columns={'created_date': 'create_date_notification'})
df_transactions = df_transactions.rename(columns={'created_date': 'create_date_transaction'})
df_users = df_users.rename(columns={'created_date': 'create_date_user'})

display(df_devices, df_notifications, df_transactions, df_users)
```

```python
df_users['num_successful_referrals'].sum()
```

# Merge

```python
pd.set_option('display.max_columns', None)

# Empieza con df_users
df = df_users.copy()

# Merge con devices
df = df.merge(df_devices, on='user_id', how='left')
df['has_device'] = df['brand_device'].notna()

# Merge con notifications
df = df.merge(df_notifications, on='user_id', how='left')
df['has_notification'] = df['create_date_notification'].notna()

# Merge con transactions
df = df.merge(df_transactions, on='user_id', how='left')
df['has_transaction'] = df['transaction_id'].notna()


df.head(30)

df['user_id'].nunique()
```

```python
df.shape
```

```python
df.isnull().sum().sort_values(ascending=False)
df['user_id'].nunique()
```

📌 1. Identifica columnas con demasiados nulos

Estas columnas tienen más de 70% de nulos, lo que generalmente amerita eliminarlas:

```python
cols_to_drop = [
    'ea_cardholderpresence',       # 295,484 nulos
    'ea_merchant_city',            # 294,220 nulos
    'ea_merchant_country',         # 294,213 nulos
    'ea_merchant_mcc'             # 294,204 nulos
]

df = df.drop(columns=cols_to_drop)
df.isnull().sum().sort_values(ascending=False)
df['user_id'].nunique()
```

📌 2. Imputa columnas con banderas booleanas

Estas columnas tienen ~122k nulos, pero su naturaleza es binaria (0/1). Lo más probable es que los nulos signifiquen "no lo activó", así que los puedes rellenar con 0:

```python
df['attributes_notifications_marketing_push'] = df['attributes_notifications_marketing_push'].fillna(0)
df['attributes_notifications_marketing_email'] = df['attributes_notifications_marketing_email'].fillna(0)

df.isnull().sum().sort_values(ascending=False)
df['user_id'].nunique()
```

📌 3. Evalúa usuarios sin actividad

Estas columnas tienen 54,604 nulos, y todas están relacionadas con transacciones:


create_date_transaction

direction

transactions_state

amount_usd

transactions_currency

transactions_type

transaction_id

👉 Eso indica que hay usuarios sin transacciones. Solo queremos usuarios activos:

```python
# df = df[~df['create_date_transaction'].isnull()]
# df.isnull().sum().sort_values(ascending=False)

# df['user_id'].nunique()
```

📌 4. Imputa columnas de notificaciones
Estas tres columnas tienen 1,010 nulos:

status

create_date_notification

channel

reason

Rellenamos con 'unknown':

```python
df['status'] = df['status'].fillna('unknown')
df['create_date_notification'] = df['create_date_notification'].fillna('unknown')
df['channel'] = df['channel'].fillna('unknown')
df['reason'] = df['reason'].fillna('unknown')
df.isnull().sum().sort_values(ascending=False)
```

📊 4. Clustering (K-Means)

# Seguimiento del movimiento de usuarios entre clústeres con el tiempo

__Número de transacciones (num_transactions)__

```python
num_transactions = df_transactions.groupby('user_id').size().reset_index(name='num_transactions')
num_transactions
```

__Número de notificaciones (num_notifications)__

```python
num_notifications = df_notifications.groupby('user_id').size().reset_index(name='num_notifications')
num_notifications
```

__Unirlos al df__

```python
# Si aún no están en df
df = df.merge(num_transactions, on='user_id', how='left')
df = df.merge(num_notifications, on='user_id', how='left')

# Si lo necesitas, rellena los nulos con 0 (por ejemplo si algún usuario nunca hizo transacciones o recibió notificaciones)
df[['num_transactions', 'num_notifications']] = df[['num_transactions', 'num_notifications']].fillna(0)

df

df['user_id'].nunique()
```

__Agrupar por usuario y periodo de tiempo (ej. mensual)__

```python
df['month_create_date_user'] = df['create_date_user'].dt.to_period('M')

user_monthly = df.groupby(['user_id', 'month_create_date_user']).agg({
    'num_transactions': 'sum',
    'num_notifications': 'sum'
}).reset_index()

user_monthly
```

__Normalizar variables y aplicar clustering por mes__

```python
user_monthly['cluster'] = None

for period in user_monthly['month_create_date_user'].unique():
    data_period = user_monthly[user_monthly['month_create_date_user'] == period]
    
    X = data_period[['num_transactions', 'num_notifications']]
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    kmeans = KMeans(n_clusters=4, random_state=42)
    clusters = kmeans.fit_predict(X_scaled)

    user_monthly.loc[data_period.index, 'cluster'] = clusters

user_monthly
```

__Analizar el cambio de clúster por usuario__

```python
pivot = user_monthly.pivot(index='user_id', columns='month_create_date_user', values='cluster')
pivot
```

# calcular la conversión

```python
# el converted ideal debe identificar a usuarios que:

# Recibieron una notificación (status == 'SENT'),

# Hicieron una transacción (transaction_id.notnull()),

# La transacción fue después de la notificación,

# Y ocurrió en un plazo razonable (por ejemplo, 3 días).

df["converted"] = (
    (df["status"] == "SENT") &
    (df["transaction_id"].notna()) &
    (df["create_date_transaction"] > df["create_date_notification"]) &
    ((df["create_date_transaction"] - df["create_date_notification"]) <= pd.Timedelta(days=3))
)

df

df['user_id'].nunique()
```

__Agrupar por canal y motivo para calcular métricas__

```python
conversion_stats = df.groupby(['channel', 'reason']).agg(
    total_notifications=('user_id', 'count'),
    total_converted=('converted', 'sum')
).reset_index()

conversion_stats['conversion_rate'] = conversion_stats['total_converted'] / conversion_stats['total_notifications']

conversion_stats
```

__Visualizar con Seaborn__

```python
fig = px.bar(
    conversion_stats,
    x='channel',
    y='conversion_rate',
    color='reason',
    barmode='group',
    title='Tasa de conversión por canal y motivo de notificación',
    text='conversion_rate'
)
fig.update_layout(xaxis_title='Canal', yaxis_title='Tasa de conversión')
fig.show()
```

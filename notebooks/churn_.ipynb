```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
df = pd.read_csv('../data/df_.csv')

df = df.drop(columns=['Unnamed: 0'])

df
```

```python
df_con_transacciones = df[df['has_transaction'] == True]
df_con_transacciones
```

```python
# Asegúrate de que 'create_date_transaction' esté en formato datetime
df['create_date_transaction'] = pd.to_datetime(df['create_date_transaction'])

# Última transacción por usuario
ultima_txn = df.groupby('user_id')['create_date_transaction'].max().reset_index()
ultima_txn.columns = ['user_id', 'ultima_txn']

ultima_txn
```

```python
fecha_max = df['create_date_transaction'].max()
fecha_max
```

```python
# Merge con el dataframe original
df_merged = df.drop_duplicates(subset='user_id').merge(ultima_txn, on='user_id', how='left')

# Calcular días inactivo
df_merged['dias_inactivo'] = (fecha_max - df_merged['ultima_txn']).dt.days

df_merged
```

```python
# Consideramos churn si tiene más de 60 días sin transaccionar
df_merged['churned'] = df_merged['dias_inactivo'] > 60

df_merged
```
